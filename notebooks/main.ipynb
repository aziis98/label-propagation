## MNIST Dataset

Load MNIST via `sklearn.datasets.fetch_openml` and visualize a few sample digits.


```python
from sklearn.datasets import fetch_openml

X_mnist, y_mnist = fetch_openml("mnist_784", version=1, return_X_y=True, as_frame=False, parser="liac-arff")
y_mnist = y_mnist.astype(int)

print(f"X_mnist shape: {X_mnist.shape}, dtype: {X_mnist.dtype}")
print(f"y_mnist shape: {y_mnist.shape}, classes: {np.unique(y_mnist)}")

fig, axes = plt.subplots(1, 5, figsize=(10, 2.5))
for i, ax in enumerate(axes):
    ax.imshow(X_mnist[i].reshape(28, 28), cmap="gray")
    ax.set_title(f"Label: {y_mnist[i]}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```
