## Classic Example Datasets

Two spirals, two moons, and concentric circles.


```python
from sklearn.datasets import make_circles, make_moons


def make_two_spirals(n_samples=400, noise=0.35, seed=42):
    rng = np.random.default_rng(seed)
    n = n_samples // 2
    theta = np.sqrt(rng.uniform(0, 1, n)) * 3 * np.pi
    r_a = 2 * theta + np.pi
    data_a = np.column_stack([np.cos(theta) * r_a, np.sin(theta) * r_a]) + rng.normal(0, noise, (n, 2))
    data_b = np.column_stack([-np.cos(theta) * r_a, -np.sin(theta) * r_a]) + rng.normal(0, noise, (n, 2))
    X = np.vstack([data_a, data_b])
    y = np.hstack([np.zeros(n, dtype=int), np.ones(n, dtype=int)])
    return X, y


X_spirals, y_spirals = make_two_spirals(n_samples=400, noise=0.35, seed=42)
X_moons, y_moons = make_moons(n_samples=400, noise=0.08, random_state=42)
X_circles, y_circles = make_circles(n_samples=400, factor=0.45, noise=0.06, random_state=42)

datasets = [
    ("Two Spirals", X_spirals, y_spirals),
    ("Two Moons", X_moons, y_moons),
    ("Concentric Circles", X_circles, y_circles),
]

fig, axes = plt.subplots(1, 3, figsize=(12, 3.8))
for ax, (title, X, y) in zip(axes, datasets):
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", s=18, edgecolors="k", linewidths=0.3)
    ax.set_title(title)
    ax.set_aspect("equal")
plt.tight_layout()
plt.show()
```

## MNIST Dataset

Load MNIST via `sklearn.datasets.fetch_openml` and visualize a few sample digits.


```python
from sklearn.datasets import fetch_openml

X_mnist, y_mnist = fetch_openml("mnist_784", version=1, return_X_y=True, as_frame=False, parser="liac-arff")
y_mnist = y_mnist.astype(int)

print(f"X_mnist shape: {X_mnist.shape}, dtype: {X_mnist.dtype}")
print(f"y_mnist shape: {y_mnist.shape}, classes: {np.unique(y_mnist)}")

fig, axes = plt.subplots(1, 5, figsize=(10, 2.5))
for i, ax in enumerate(axes):
    ax.imshow(X_mnist[i].reshape(28, 28), cmap="gray")
    ax.set_title(f"Label: {y_mnist[i]}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```
